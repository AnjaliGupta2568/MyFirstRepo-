# #1.Array Manipulation Function

```python
import numpy as np
```

```python
a = np.array([1,2,3,4])           #reshape
reshaped = np.reshape(a,(1,4))
print("Reshaped array :",reshaped)
```

```python
a = np.array([1,2,3,4])
reshaped = np.reshape(a,(4,1))
print("Reshaped array :\n",reshaped)
```

```python
b = np.array([[4,5,9],[6,7,8]])       #ravel
flattened = np.ravel(b)
print("Flattened array :",flattened)
```

```python
c =  np.array([[4,5,9],[6,7,8]])       #transpose
transposed = np.transpose(c)
print("Transposed array :\n",transposed)
```

```python
a1 = np.array([1,2])
b1 = np.array([3,4])
c1 = np.array([5,6])
stacked = np.vstack([a1,b1,c1])      #vstack
print("Stacked arrays :\n",stacked)
```

# 6. Boolean & Logical Functions

```python
# check if all elements are True
#all
logical_test  = np.array([True,False,True])
all_true = np.all(logical_test)
print("All elements true:",all_true)
```

# 7.Set Operation

```python
set_a = np.array([1,7,3,9])
set_b = np.array([2,9,3,0])
Intersection = np.intersect1d(set_a,set_b)      #intersection
print('Intersection of a and b:',Intersection)
```

```python
set_a = np.array([1,7,3,9])
set_b = np.array([2,9,3,0])
Union = np.union1d(set_a,set_b)      #union
print('Union of a and b:',Union)
```

# 9.Other functions

```python
arr = np.array([[1,2,3],[4,5,7]])   
copied_arr = np.copy(arr)     #create a copy of array a
print('copied array =\n',copied_arr)
```

```python
array_size_in_bytes = arr.nbytes  # size in bytes of array
print('Size of arr in bytes =',array_size_in_bytes)
```

```python
shared = np.shares_memory(arr,copied_arr)     #check if arrays share memory
print('Do arr and copied_arr share memory?',shared)
```

# 4.if-elif-else

```python
age = 35
if age<=12:
    print("Child")
elif age<=19:
    print("Teenager.")
elif age <=35:
    print("Young adult.")
else :
    print("adult")
    
```

```python
number = int(input("Enter a number: "))

if number > 0:
    print("The number is positive.")
elif number == 0:
    print("The number is zero.")
else:
    print("The number is negative.")
```

```python
score = int(input("Enter your score: "))

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

# 5.Nested-if

```python
age=75
is_member = True

if age>=60:
    if is_member:
        print('30% senior discount')
    else:
        print('20% senior discount')
else:
    print('Not eligible for a senior discount.')
```

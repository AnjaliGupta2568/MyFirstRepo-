# 2.Right-angled triangle

```python
for i in range(6):
    for j in range(i+1):
        print('#',end=' ')
    print()    
```

```python
for i in range(1,6):
    print("# "*i)
```

# 3.Inverted Right angled triangle

```python
for i in range(4):
    for j in range(4-i):
        print('#', end="  ")
    print()
```

```python
for i in range(1,5):
    print(" # "*(5-i))
```

# 4.Pyramid 

```python
for i in range(1,6):
    print(''*(5-i) + ' # '*(2*i-1) )
```

```python
for i in range(1,6):
    for j in range(1,10):
        if ((j>=6-i) and (j<=4+i)):
            print(' * ',end=' ')
        else:
            print('   ',end=' ')
            
    print() 
       
```

# 5.Inverted Pyramid

```python
for i in range(5,0,-1):
    print(''*(5-i)+' # '*(2*i-1))
```

```python
for i in range(1,6):
    for j in range(1,10):
        if ((j>=i) and (j<=10-i)):
            print(' * ',end=' ')
        else:
            print('   ',end=' ')
            
    print() 
```

# 6.Diamond Pattern

```python
for i in range(1,6):
    for j in range(1,10):
        if ((j>=6-i) and (j<=4+i)):
            print(' * ',end=' ')
        else:
            print('   ',end=' ')
            
    print() 

for i in range(2,6):
    for j in range(1,10):
        if ((j>=i) and (j<=10-i)):
            print(' * ',end=' ')
        else:
            print('   ',end=' ')
            
    print()     
```

# 7.Hallow Square Pattern

```python
for i in range(5):
    for j in range(5):
        if(i==4 or i==0 or j==4 or j==0):
             print('*',end =' ')
        else:
             print(' ',end =' ')
            
    print()
```

# 8.Full Square Pattern

```python
for i in range(5):
    for j in range(5):
             print('*',end =' ')
            
    print()
```

# 9.Right angled triangle(number pattern)

```python
for i in range(5):
    for j in range(i):
        print(i,*(' '),end = ' ')
    print()
```

# 10.Inverted Right angled triangle(number pattern)

```python
for i in range(6):
    for j in range(5-i):
        print(i,*(' '),end = ' ')
    print()    
```

# 11.Hallow Right angled triangle

```python
for i in range(1,6):
    for j in range(1,i+1):
        if (j==1 or i==5 or j==i):
            print(' * ',end='')
        else:
            print('   ',end='')
            
    print() 
```

# 12.Hallow Pyramid pattern

```python
for i in range(1,6):
    for j in range(1,10):
        if(i==5 or j==6-i or j==4+i):
            print(' * ',end ='')
        else:
            print('   ',end ='')
            
    print()
      
```

# 13.Hallow Diamond Pattern

```python
for i in range(1,5):
    for j in range(1,8):
        if(j==5-i or j==3+i):
            print(' * ',end ='')
        else:
            print('   ',end ='')

    print()   

for i in range(2,5):
    for j in range(1,8):
        if(j==i or j==8-i):
            print(' * ',end='')
        else:
            print('   ',end='')

    print()
```

# 14.Hallow Diamond Pattern(number pattern)

```python
for i in range(1,5):
    for j in range(1,8):
        if(j==5-i or j==3+i):
            print(f' {i} ',end ='')
        else:
            print('   ',end ='')

    print()   

for i in range(2,5):
    for j in range(1,8):
        if(j==i or j==8-i):
            print(f' {i} ',end='')
        else:
            print('   ',end='')

    print()
```

# 15.Butterfly pattern

```python
for i in range(1,5):
    for j in range(1,8):
        if((j>=1 and j<=i) or (j>=8-i and j<=7)):
            print(' * ',end='')
        else:
            print('   ',end='')

    print() 
    
for i in range(1,5):
    for j in range(1,8):
        if((j>=1 and j<=5-i) or (j>=3+i and j<=7)):
            print(' * ',end='')
        else:
            print('   ',end='')

    print()    
        
```

# 19.Left Aligned Pyramid

```python
for i in range(6):
    for j in range(i+1):
        print('*',end=' ')
    print() 
```

# 20.Right Aligned Pyramid

```python
for i in range(1,6):
    for j in range(1,6):
        if(j>=6-i and j<=5):
            print(' * ',end=' ')
        else:
            print('   ',end=' ')

    print() 
```

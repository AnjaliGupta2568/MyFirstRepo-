# Nested while-loop

```python
i = 1

while i<=5:
    print("data science")
    j=1
    while j<=4:
        print('technology')
        j = j + 1

    i = i + 1
    print()
```

```python
#  when we mention end then new line will not create
i = 1

while i<=5:
    print("data science",end = " ")
    j=1
    while j<=4:
        print('technology',end = " ")
        j = j + 1

    i = i + 1
    print()
```

# while-else

```python
i = 10

while i <= 4 :
    j = 1
    while  j <= 3 :
        print(i*j, end=" ")
        j += 1
         
    print()
    i += 1
    
else:
    print('condition is not match')
```

```python
# Check if a number is prime
num = 17
i = 2

while i < num:
    if num % i == 0:
        print(f"{num} is not a prime number.")
        break
    i += 1
else:
    print(f"{num} is a prime number.")
```

# 2.For loop

```python
name = 'hello'    #intialization

for i in name:
    print(i)
```

```python
#Iterating Over List, Tuple, String and Dictionary Using for Loops 
l = ["geeks", "for", "geeks"]    #list
for x in l:
    print(x)
    
tup = ("geeks", "bye", "geeks")   #tuple
for x in tup:
    print(x)
    
s = "hello"      #string
for x in s:
    print(x)
    
d = dict({'x':123, 'y':354})     #dictionary
for x in d:
    print("%s  %d" % (x, d[x]))
    
s = {10, 30, 20}              #set
for x in s:
    print(x),
```

# nested-for loop

```python
for i in range(4):
    for j in range(3):
        print(i)
```

```python
for i in range(4):
    for j in range(3):
        print(i,end=" ")
```

```python
for i in range(1, 5):
    for j in range(i):
        print(i, end=' ')
    print()
```

```python
for i in range(1, 6):       # Outer loop for rows
    for j in range(1, 6):     # Inner loop for columns
        print(i * j, end=" ")  
    print()                         # Newline after each row
```

# for-else

```python
nums = [12,18,21,26]

for num in nums:
    if num % 5 == 0:
        print(num)
        #break

else:
    print('number not found')
```

```python
nums = [12,18,21,26,15]

for num in nums:
    if num % 3 == 0:
        print(num)
        break

else:
    print('number not found')
```

# 3.Break,Pass,Continue

```python
for i in range(1,11):
    if i == 4: 
       break 
    print(i)
```

```python
for i in range(1,11):
    if i == 6: 
       continue 
    print(i)
```

```python
for i in range(1,11):
    pass                #we use pass statement in Python to write empty loops
```

```python
for i in range(1,21):
    
    if i%3 == 0:
        continue
    print(i)
print('end')
```

```python
for i in range(1,30): 
    
    if i%3 == 0 or i%5 == 0:
        
        continue
    print(i,end = " ")
```

```python
nums = [12,18,21,26]

for num in nums:
    if num % 5 == 0:
        print(num)
        break
```
